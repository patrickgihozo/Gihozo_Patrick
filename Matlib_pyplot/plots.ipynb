```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

fig, ax = plt.subplots(figsize=(8, 5))

ax.plot(x, y, marker='o', color='blue', label='Sales')

ax.set_title("Sales Growth")


ax.set_xlabel("Month")
ax.set_ylabel("Sales")

ax.legend()
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

fig,ax = plt.subplots()

plt.bar(x, y)
plt.title("Basic line graph")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]


fig, ax = plt.subplots(figsize=(8, 5))
ax.scatter(x, y, color='green', s=100, label='Sales')

ax.set_title("Scatter Plot Example")
ax.set_xlabel("Month")
ax.set_ylabel("Sales")
ax.legend()


plt.show()
```

```python
teams =["Argentina","Brazil","France","Germany","Paraguay"]
wins =[2,4,3,1,1]

fig,ax =plt.subplots(figsize=(6,6))
ax.pie( wins,labels=teams,
    autopct="%1.1f%%",
    startangle=90 )
ax.set_title("World cup winners")
ax.legend(teams,title="Teams")
plt.show
```
