```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv("churn_dataset.csv")
print(df)
print(df.info())
print(df.describe())
print(df.columns)
print(df.shape)
print(df.tail())
print(df.head())
```

```python
x = df[["Age","MonthlyBill","Complaints"]]
y = df["Churn"]

x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42
)
```

```python
model = LogisticRegression() #Create logistic regression object
model.fit(x_train, y_train)#Fit function trains model object created on data
print(model.fit(x_train, y_train))#Fit function returns LogisticRegression that shows a learned model.
```

```python
predictions = model.predict(x_test)
accuracy = accuracy_score(y_test,predictions)
print("Predictions :",predictions)
print("Actual:",list(y_test))
print(f"Accuracy  = {accuracy*100:.2f}%")
```
